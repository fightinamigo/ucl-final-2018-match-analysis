```python
from statsbombpy import sb
import pandas as pd

events = sb.events(match_id=18245)
shots = events[events["type"] == "Shot"].copy()
shots[["x", "y"]] = shots["location"].apply(pd.Series)
shots[["team", "player", "minute", "x", "y", "shot_statsbomb_xg", "shot_outcome"]].head()
```

```python
summary = shots.groupby("team").agg(
    shots=("player", "count"),
    on_target=("shot_outcome", lambda s: s.isin(["Goal", "Saved"]).sum()),
    goals=("shot_outcome", lambda s: (s == "Goal").sum()),
    xG=("shot_statsbomb_xg", "sum"),
).round(2)
summary
```

Both teams took 14 shots, but Real Madrid created slightly better chances (1.68 xG vs 1.43) and made them count, scoring 3 goals to Liverpool’s 1.

```python
import matplotlib.pyplot as plt
from mplsoccer import Pitch

pitch = Pitch(pitch_type="statsbomb")
fig, ax = pitch.draw(figsize=(10, 7))

for team, color in [("Real Madrid", "white"), ("Liverpool", "red")]:
    t = shots[shots["team"] == team]
    pitch.scatter(t.x, t.y, s=t.shot_statsbomb_xg * 1500, c=color,
                  edgecolors="black", alpha=0.8, ax=ax, label=team)

ax.legend()
ax.set_title("Shots by xG: Real Madrid vs Liverpool")
plt.show()
```

Liverpool’s best chances (the biggest red dots) came close to goal, but Real Madrid’s shots were spread across the box and they converted more of them.

```python
events = sb.events(match_id=18245)
p = events[(events["type"] == "Pass") & (events["player"] == "Toni Kroos")].copy()
p[["x", "y"]] = p["location"].apply(pd.Series)
p[["end_x", "end_y"]] = p["pass_end_location"].apply(pd.Series)

pitch = Pitch(pitch_type="statsbomb")
fig, ax = pitch.draw(figsize=(10, 7))
pitch.arrows(p.x, p.y, p.end_x, p.end_y, ax=ax, width=2, color="blue")
ax.set_title("Toni Kroos passes")
plt.show()
```

Toni Kroos’s passes are concentrated in the middle of the pitch, showing he acted as Madrid’s playmaker, distributing the ball from deep rather than getting into the box.


Conclusion: Real Madrid won because they were more efficient, not because they dominated. Both teams took 14 shots, but Madrid had more shots on target (6 vs 3) and scored 3 goals to Liverpool’s 1. Their xG was only slightly higher (1.68 vs 1.43), so finishing made the difference.
